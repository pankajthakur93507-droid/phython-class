```python
def add_num(*args):
  sum=0
  for i in args:
    sum=sum+i
  return sum
add_num(20,30)
```

```python
def add_num(**kwargs):
  sum=0
  for key,value in kwargs.items():
    sum=sum+value
  print(sum)
add_num(a=10,b=30)
```

```python
def fab(n):
  if n==1:
   return 0
  if n==2:
    return 1
  return fab(n-1)+fab(n-2)
fab(10)
```

```python
def fact(n):
  if n==0:
    return 1
  return n*fact(n-1)
fact(5)
```

```python
import csv
customer,name,gender, state ,product,amount,date=[],[],[],[],[],[],[]
with open("data.csv","r") as file:
  reader=csv.reader(file)
  header = next(reader)
  for row in reader:
    customer.append(row[1])
    gender.append(row[2])
    state.append(row[3])
    product.append(row[4])
    amount.append(int(row[5]))
    date.append(row[6])
print(customer)
print(gender)
print(state)
print(product)
print(amount)
print(date)

u_n=[]
statewise_sale=[]
for i in state:
  if i not in u_n:
    u_n.append(i)
    statewise_sale.append(0)
print(u_n)
print(statewise_sale)

for i in range(len(state)):
  index=u_n.index(state[i])
  statewise_sale[index] += amount[i]
print(statewise_sale)
import matplotlib.pyplot as plt
plt.bar(u_n,statewise_sale)
plt.show()

import pandas as pd
df=pd.read_csv("data.csv")
print(df)

df.groupby("State")["Amount"].sum().plot(kind="pie")


```
